```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Loaded Successfully")
```

```python
import pandas as pd

data = pd.read_csv("ChatBot_Dataset.csv")

if "Unnamed: 0" in data.columns:
    data = data.drop(columns=["Unnamed: 0"])

data.head()
```

```python
import string
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words("english"))

def preprocess(text):
    text = str(text).lower()
    text = "".join(ch for ch in text if ch not in string.punctuation)
    words = text.split()
    words = [lemmatizer.lemmatize(w) for w in words if w not in stop_words]
    return " ".join(words)

data["processed_question"] = data["question"].apply(preprocess)

print(data.head())
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Loaded Successfully")
```

```python
question_embeddings = model.encode(
    data["processed_question"].tolist(),
    convert_to_tensor=False
)

print("Embeddings Created Successfully")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def chatbot(user_input):

    # User input preprocess
    processed = preprocess(user_input)

    # Convert user question into embedding
    user_embedding = model.encode([processed], convert_to_tensor=False)

    # Calculate similarity
    similarity = cosine_similarity(user_embedding, question_embeddings)

    # Best matching question
    best_match = similarity.argmax()

    # Return corresponding answer
    return data.iloc[best_match]["answer"]
```

```python
print("=" * 50)
print("🤖 AI Chatbot")
print("=" * 50)
print("Type 'exit' to quit.\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye! 👋")
        break

    response = chatbot(user_input)
    print("Bot:", response)
    print()
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tokenizer = AutoTokenizer.from_pretrained(model_name)

llm = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

generator = pipeline(
    "text-generation",
    model=llm,
    tokenizer=tokenizer
)

print("LLM Loaded Successfully")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def rag_chatbot(user_input):
    greetings = {
        "hi": "Hello! 👋 How can I help you today?",
        "hello": "Hello! 👋 How can I help you today?",
        "hey": "Hi! 👋 How can I help you today?",
        "bye": "Goodbye! 👋 Have a great day!",
        "thanks": "You're welcome! 😊",
        "thank you": "You're welcome! 😊"
    }

    if user_input.lower().strip() in greetings:
        return greetings[user_input.lower().strip()]
    processed = preprocess(user_input)


    user_embedding = model.encode([processed], convert_to_tensor=False)

    similarity = cosine_similarity(user_embedding, question_embeddings)

    idx = similarity.argmax()

    score = similarity[0][idx]
    if score < 0.40:
        return "Sorry, I couldn't find a relevant answer in my knowledge base."

    # Retrieve context
    context = str(data.iloc[idx]["answer"])

    prompt = f"""
You are a helpful AI assistant.

Answer ONLY using the given context.
If the answer is not available, say "I don't know."

Context:
{context}

Question:
{user_input}

Answer:
"""

    output = generator(
        prompt,
        max_new_tokens=80,
        do_sample=False,
        return_full_text=False
    )

    answer = output[0]["generated_text"].strip()

    return answer
```

```python
print("\n" + "="*65)
print("🤖        SMART GENERATIVE AI CHATBOT")
print("="*65)
print("💡 Ask me anything from my knowledge base.")
print("❌ Type 'exit' anytime to quit.")
print("-"*65)

while True:

    question = input("\n👤 You : ")

    if question.lower().strip() == "exit":
        print("\n🤖 Bot : Goodbye! 👋 Have a great day!")
        print("="*65)
        break

    print("\n🤖 Thinking... ⏳")

    answer = rag_chatbot(question)

    print("\n" + "-"*65)
    print("🤖 AI Assistant:")
    print(answer)
    print("-"*65)
```
